```python
x = df['bmi'].values
y = df['claim'].values
```

```python
x_train , x_test , y_train, y_test = train_test_split(x, y , test_size = 0.2, random_state = 2)
```

# Model

```python
class GDRegressor :

  def __init__(self , learning_rate = 1 , epochs = 100):
    self.m = 1
    self.b = 0 
    self.lr = learning_rate
    self.epochs = epochs


  def fit(self , x_train , y_train):

    for i in range(self.epochs):
      # b_slop = -2 * np.sum( y_train - self.m * x_train- self.b)
      m_slop = -2 * np.sum( (y_train - self.m * x_train - self.b)* x_train  )
    
      print(m_slop.shape)
      print(m_slop)
        
      # self.b = self.b - (self.lr * b_slop)
      self.m = self.m - (self.lr * m_slop)
      
        

  def predict (self , x_test ):
    return self.m * x_test + self.b
```

```python
pd.Series(y_train).isnull().sum()
```

```python
# object 
model = GDRegressor(epochs = 10)
# fit
model.fit(x_train,y_train)
# predict
y_pred = model.predict(x_test)
print(y_pred)
```

```python
from sklearn.metrics import r2_score
```

```python
r2_score(y_test , y_pred)
```
